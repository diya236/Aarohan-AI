### Observation :

I have seen that there are many students who have physical activity time >10 hours which is practically immpossible so i am trying to find that if the sum of all study hours are = 24 or not so we can predict that datset is basically design in such a way that the sum will become 24 

```python
hour_cols = ["Study_Hours_Per_Day", "Extracurricular_Hours_Per_Day",
             "Sleep_Hours_Per_Day", "Social_Hours_Per_Day",
             "Physical_Activity_Hours_Per_Day"] #takes all houes rows
df["Total_Hours"] = df[hour_cols].sum(axis=1) #sum them
df["Total_Hours"].describe()
```

### Observation
I have seen that yesss it is designed in such a way that sum becomes 24 hours now the next cell will give the rows where physical activity hours > 10

```python
df[df["Physical_Activity_Hours_Per_Day"] > 10]
```
