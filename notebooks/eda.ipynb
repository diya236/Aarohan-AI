### Read Data

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("../dataset/cleaned_student_data.csv")
df.head()
```

### Avg Calculation

for ML model we can use avg just to determine that if the student is above avg then academics is going good but if below avg where he or she is lacking so it's used for performance observation

```python
df["GPA"].mean()
```

### Avg Study Hours

basically it shows student's efforts if a student is studying for more then avg time then how much it effects to result if not then waht happens

```python
df["Study_Hours_Per_Day"].mean()
```

### Stress Count

```python
df["Stress_Level"].value_counts()
```

### Stress bar chart

```python
order = ["Low", "Moderate", "High"]
sns.countplot(x="Stress_Level", data=df, order=order)
plt.title("Stress Level Distribution")
plt.xlabel("Stress Level")
plt.ylabel("Number of Students")
plt.show()
```

### Study hours vs gpa

```python
sns.scatterplot(x="Study_Hours_Per_Day", y="GPA", data=df, alpha=0.5)
plt.title("Study Hours vs GPA")
plt.show()

df["Study_Hours_Per_Day"].corr(df["GPA"])
```

### Sleep hours vs GPA

```python
sns.scatterplot(x="Sleep_Hours_Per_Day", y="GPA", data=df, alpha=0.5)
plt.title("Sleep Hours vs GPA")
plt.show()

df["Sleep_Hours_Per_Day"].corr(df["GPA"])
```

### physical Activity vs GPA

```python
sns.scatterplot(x="Physical_Activity_Hours_Per_Day", y="GPA", data=df, alpha=0.5)
plt.title("Physical Activity vs GPA")
plt.show()

df["Physical_Activity_Hours_Per_Day"].corr(df["GPA"])
```

### Correlation heatmap

```python
corr = df.drop(columns=["Student_ID"]).corr(numeric_only=True)

plt.figure(figsize=(9, 6))
sns.heatmap(corr, annot=True, fmt=".2f", cmap="coolwarm")
plt.title("Correlation Heatmap")
plt.show()
```
